# Defining Agents. 
- Struct definitions for all persons and households in the model
- Specify properties of persons and households.
- Define constructors to initialize structs.

1. Agents' possible status and properties.

- Define a Status [enumerated type](https://riptutorial.com/julia-lang/example/23840/defining-an-enumerated-type) with three states: potential, intention, migrant.
- Generate a *Person* struct.
    - [Working definition of struct](https://syl1.gitbook.io/julia-language-a-concise-tutorial/language-core/custom-types): "Structures (previously known in Julia as "Types") are, for the most (see later for the difference), what in other languages are called classes, or "structured data": they define the kind of information that is embedded in the structure, that is a set of fields (aka "properties" in other languages), and then individual instances (or "objects") can be produced each with its own specific values for the fields defined by the structure." 
    - The struct *Person* has 8 properties, each of them consisting of a different *type*. A person has:
        - Unique id
        - Status
        - Family
        - Community
        - Savings
        - Income
        - Potential gain & losses from migration

```julia
@enum Status potential intention migrant
mutable struct Person
   id :: Int64
   status :: Status
   family :: Vector{Person} #vector of persons.
   community :: Vector{Person}
   totsav :: Float64 #Cumulative savings
   ainc :: Float64 #Agents' income in each step. It represents a fixed income multiplied by a random proportion.
   gain :: Float64
   loss :: Float64
end
```

Creating *Person* constructor initializes the struct. 
- We also specify the default person initialization for each of *Person* properties.
- Last line checks the methods associated with the constructor *Person*. 

```julia

Person() = Person(1, potential, [], [], 0, 0, 1, 1) # Default person is potential and has no contacts
Person(id :: Int64) = Person(id, potential, [], [], 0, 0, 1, 1)
Person(state :: Status) = Person(1, state, [], [], 0, 0, 1, 1) # Default person has no contacts and 0 income.
Person(totsav, ainc) = Person(1, potential, [], [], totsav, ainc, 1, 1)
Person(gain :: Float64) = Person(1, potential, [], [], 0, 0, gain, 1)
Person(loss :: Float64) = Person(1, potential, [], [], 0, 0, 1, loss)
methods(Person) # Check the constructors created for Person.
```

We now generate a *household* status, struct, and constructors.
- *Household* structs have
    - Members
    - Status
    - Vulnerability level.

```julia
# Household struct
@enum statusHH potential_hh intention_hh

mutable struct Household
  members :: Vector{Person}
  status_hh :: statusHH
  vuln :: Float64 #Household vulnerability
end

Household() = Household([], potential_hh, 0)
Household(status_hh :: statusHH) = Household([], status_hh, 0)
Household(vuln :: Float64) = Household([], potential_hh, vuln)
```
